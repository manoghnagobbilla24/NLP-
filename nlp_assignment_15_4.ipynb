Part A: ELMo + Naive Bayes


```python
import tensorflow_hub as hub
import tensorflow as tf
import numpy as np
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
```

```python
sentences = ["Win money now", "Claim your prize", "Hello how are you",
             "Let's meet tomorrow", "Free lottery ticket", "Are you coming today"]

labels = np.array([1, 1, 0, 0, 1, 0])
```

```python
elmo = hub.load("https://tfhub.dev/google/elmo/3")
```

```python
embeddings = elmo.signatures["default"](tf.constant(sentences))["elmo"]
```

```python
embeddings = elmo.signatures["default"](tf.constant(sentences))["elmo"]
sentence_embeddings = tf.reduce_mean(embeddings, axis=1).numpy()
```

```python
model_elmo = GaussianNB()
model_elmo.fit(sentence_embeddings, labels)
```

```python
pred_elmo = model_elmo.predict(sentence_embeddings)
print("ELMo Accuracy:", accuracy_score(labels, pred_elmo))
```

Part B: BERT + Naive Bayes


```python
import tensorflow_hub as hub
import tensorflow_text as text
import numpy as np
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
```

```python
sentences = ["Win money now", "Claim your prize", "Hello how are you",
             "Let's meet tomorrow", "Free lottery ticket", "Are you coming today"]

labels = np.array([1, 1, 0, 0, 1, 0])
```

```python
preprocess = hub.load("https://tfhub.dev/tensorflow/bert_en_uncased_preprocess/3")
bert = hub.load("https://tfhub.dev/tensorflow/bert_en_uncased_L-12_H-768_A-12/4")
```

```python
inputs = preprocess(sentences)
```

```python
outputs = bert(inputs)
sentence_embeddings = outputs["pooled_output"].numpy()
```

```python
model_bert = GaussianNB()
model_bert.fit(sentence_embeddings, labels)
```

```python
pred_bert = model_bert.predict(sentence_embeddings)
print("BERT Accuracy:", accuracy_score(labels, pred_bert))
```

Part C: Comparison


```python
from sklearn.metrics import accuracy_score

# --- Accuracy ---
elmo_acc = accuracy_score(labels, pred_elmo)
bert_acc = accuracy_score(labels, pred_bert)

# --- Feature Size ---
elmo_feature_size = sentence_embeddings.shape[1]   # from ELMo
bert_feature_size = 768  # BERT pooled output size

# --- Print Results ---
print("=== Comparison between ELMo and BERT ===\n")

print("Accuracy:")
print("ELMo Accuracy:", elmo_acc)
print("BERT Accuracy:", bert_acc)

print("\nFeature Size:")
print("ELMo Feature Size:", elmo_feature_size)
print("BERT Feature Size:", bert_feature_size)

print("\nObservations:")
if bert_acc >= elmo_acc:
    print("- BERT performs better than ELMo in accuracy.")
else:
    print("- ELMo performs better than BERT.")

print("- BERT captures context bidirectionally.")
print("- ELMo uses LSTM-based embeddings.")
print("- BERT requires more computational power.")
print("- ELMo is faster but less powerful.")

print("\nFinal Conclusion:")
print("BERT outperforms ELMo in terms of accuracy and contextual understanding.")
```
